# 📌 Probability: Definition, Formula, and Example

## 🔹 Definition of Probability
Probability is the **measure of how likely an event is to occur**. It is expressed as a number between **0 and 1**, where:  
- **0** means the event will **never happen**.  
- **1** means the event will **always happen**.  
- **A probability closer to 1** means the event is **more likely**.  

---

## 🔹 Formula for Probability
The probability of an event **A** occurring is given by:  

\[
P(A) = \frac{\text{Number of favorable outcomes}}{\text{Total number of possible outcomes}}
\]

Where:  
- **\( P(A) \)** → Probability of event **A**  
- **Favorable outcomes** → Outcomes where **A happens**  
- **Total outcomes** → All possible outcomes in the sample space  

---

## 🔹 Example 1: Tossing a Coin
**Problem:** What is the probability of getting heads when tossing a fair coin?  
- There are **2 possible outcomes**: **Heads (H) or Tails (T)**.  
- The number of **favorable outcomes** (getting heads) = **1**.  
- Using the formula:  

\[
P(H) = \frac{1}{2} = 0.5
\]  

**Answer:** The probability of getting heads is **0.5 (50%)**.

---

## 🔹 Example 2: Rolling a Die
**Problem:** What is the probability of rolling a **4** on a fair 6-sided die?  
- There are **6 possible outcomes**: {1, 2, 3, 4, 5, 6}.  
- The number of **favorable outcomes** (rolling a **4**) = **1**.  
- Using the formula:  

\[
P(4) = \frac{1}{6} \approx 0.1667
\]  

**Answer:** The probability of rolling a 4 is **0.1667 (16.67%)**.

---

## 🔹 Types of Probability
1. **Theoretical Probability** → Based on **mathematical calculations** (e.g., rolling dice).  
2. **Experimental Probability** → Based on **real-world experiments** (e.g., flipping a coin 100 times).  
3. **Conditional Probability** → Probability of an event **given another event has already happened**.  
4. **Bayesian Probability** → Uses prior knowledge to update probability estimates.
