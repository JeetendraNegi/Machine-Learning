```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# load the dataset

dataset = pd.read_csv('grocery_item_transactions.csv')
dataset.head(3)
```

```python
# now create a new dataset (list of list) that only contains items not NaN

market = []

for i in range(0, dataset.shape[0]):
    cust = []
    for j in dataset.columns:
        if type(dataset[j][i]) == str:
            cust.append(dataset[j][i])
    market.append(cust)
```

```python
# now get the most buy items

import collections
```

```python
items = []
for i in market:
    for j in i:
        items.append(j)

collections.Counter(items)
```

```python
# now we need to figure out the association

from mlxtend.preprocessing.transactionencoder import TransactionEncoder
```

```python

tr = TransactionEncoder()
tr.fit(market)
```

```python
tr.transform(market)
```

```python
df = pd.DataFrame(tr.transform(market), columns = tr.columns_)
```

```python
# now use Apriori Association

from mlxtend.frequent_patterns import apriori
```

```python
apriori(df, min_support=0.05, use_colnames=True, max_len=3).sort_values(by=["support"])
```

```python
# now try to remove the max_length to check if we get any other association

pd.set_option('display.max_rows', None)
apriori(df, min_support=0.05, use_colnames=True).sort_values(by=["support"])
```
