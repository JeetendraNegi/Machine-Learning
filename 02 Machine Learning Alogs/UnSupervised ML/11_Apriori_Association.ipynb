# 🛒 Apriori Algorithm in Association Rule Learning

## 📘 Definition

**Apriori** is an algorithm used in **association rule learning** to find patterns or relationships between items in a dataset.  
It helps answer questions like:

> "If a customer buys X and Y, are they likely to buy Z?"

It works by identifying **frequent itemsets** (groups of items that appear often together) and then generating rules from them.

---

## 📊 Key Metrics

### ✅ Support
- How often an item or group of items appears in all transactions.
- Formula:

$[
\text{Support}(A) = \frac{\text{Number of transactions containing A}}{\text{Total transactions}}
$]

---

### ✅ Confidence
- How often item **B** is bought **when item A is bought**.
- Formula:

$[
\text{Confidence}(A \rightarrow B) = \frac{\text{Support}(A \cup B)}{\text{Support}(A)}
$]

---

### ✅ Lift
- Tells how much more likely B is to be bought when A is bought, compared to random chance.
- Formula:

$[
\text{Lift}(A \rightarrow B) = \frac{\text{Confidence}(A \rightarrow B)}{\text{Support}(B)}
$]

---

## 🔁 Step-by-Step Working of Apriori

### Step 1: Set minimum support and confidence
- Choose thresholds (e.g., support ≥ 0.5 and confidence ≥ 0.6)

### Step 2: Find frequent itemsets
- Identify combinations of items that meet the **minimum support**.

### Step 3: Generate rules
- From frequent itemsets, generate rules (e.g., A → B)

### Step 4: Filter rules
- Keep rules that meet the **confidence** and **lift** thresholds.

---

## 🧠 Simple Example

### Dataset of Transactions

| Transaction ID | Items                     |
|----------------|---------------------------|
| 1              | Bread, Milk               |
| 2              | Bread, Diaper, Beer, Eggs |
| 3              | Milk, Diaper, Beer, Coke  |
| 4              | Bread, Milk, Diaper, Beer |
| 5              | Bread, Milk, Diaper, Coke |

---

### Step 1: Support Calculation

Let’s calculate support for a few items/itemsets:

- Support(Bread) = 4/5 = 0.8  
- Support(Milk) = 4/5 = 0.8  
- Support(Bread ∩ Milk) = 3/5 = 0.6  
- Support(Diaper ∩ Beer) = 3/5 = 0.6  

---

### Step 2: Confidence Calculation

Let’s calculate **Confidence(Bread → Milk)**:

$[
\text{Confidence} = \frac{\text{Support}(Bread ∩ Milk)}{\text{Support}(Bread)} = \frac{0.6}{0.8} = 0.75
$]

So if someone buys bread, there is a **75% chance** they will also buy milk.

---

### Step 3: Lift Calculation

Let’s calculate **Lift(Bread → Milk)**:

$[
\text{Lift} = \frac{\text{Confidence}(Bread → Milk)}{\text{Support}(Milk)} = \frac{0.75}{0.8} = 0.9375
$]

Because **Lift < 1**, this means buying Bread slightly reduces the chance of buying Milk — not a strong association.

---

## ✅ Final Observations

| Rule              | Support | Confidence | Lift    | Meaning                      |
|-------------------|---------|------------|---------|------------------------------|
| Bread → Milk      | 0.6     | 0.75       | 0.9375  | Moderate relationship        |
| Diaper → Beer     | 0.6     | 0.75       | 1.25    | Strong positive association  |

---

## 📌 Summary

- Apriori finds frequent itemsets based on **support**.
- It then generates **rules** and filters them using **confidence** and **lift**.
- Good rules have:
  - **High support** (happens often)
  - **High confidence** (strong link)
  - **Lift > 1** (more than chance)

---

## 🎯 Use Cases

- Market basket analysis (e.g., groceries, e-commerce)
- Recommender systems (e.g., Amazon: "Frequently bought together")
- Web click behavior patterns
