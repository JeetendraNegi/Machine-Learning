```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# load the dataset

dataset = pd.read_csv('grocery_item_transactions.csv')
dataset.head(3)
```

```python
# now create a new dataset (list of list) that only contains items not NaN

market = []

for i in range(0, dataset.shape[0]):
    cust = []
    for j in dataset.columns:
        if type(dataset[j][i]) == str:
            cust.append(dataset[j][i])
    market.append(cust)
```

```python
# now get the most buy items

import collections
```

```python
items = []
for i in market:
    for j in i:
        items.append(j)

collections.Counter(items)
```

```python
# now we need to figure out the association

from mlxtend.preprocessing.transactionencoder import TransactionEncoder
```

```python
tr = TransactionEncoder()
tr.fit(market)
```

```python
tr.transform(market)
```

```python
df = pd.DataFrame(tr.transform(market), columns = tr.columns_)
```

```python
# import fp growth
from mlxtend.frequent_patterns import fpgrowth
```

```python
fpgrowth(df, min_support=0.05, use_colnames=True, max_len=3).sort_values(by=['support'], ascending=False)
```

```python
# now check the same without the max length
pd.set_option('display.max_rows', None)
fpgrowth(df, min_support=0.05, use_colnames=True).sort_values(by=['support'], ascending=True)
```

```python
# now display all the values which are above 0.05
fpgrowth(df, min_support=0.05, use_colnames=True).sort_values(by=['support'], ascending=True).query('support > 0.05')
```
