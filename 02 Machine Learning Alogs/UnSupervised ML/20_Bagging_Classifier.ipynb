# Bagging Classifier & Random Forest Classifier

````python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
````

````python
# prepare the dataset using sklearn

x, y = make_moons(n_samples=1000, noise=0.1, random_state=42)

dataset = pd.DataFrame(x, columns=['x1', 'x2'])
dataset['y'] = y

dataset.head(3)
````

````python
# now plot the scatter plot to see the data

plt.figure(figsize=(10,5))
sns.scatterplot(data=dataset, x='x1', y='x2', hue='y')
plt.show()
````

````python
# input and output variables

x = dataset.iloc[:, :-1]
y = dataset['y']
````

````python
# train test and split the dataset

from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
````

````python
# now use Bagging Classifier

from sklearn.ensemble import BaggingClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
````

## Use the Estimator to SVC()

````python
bg = BaggingClassifier(
    estimator= SVC(),
    n_estimators=100,
    random_state=42
)

bg.fit(x_train, y_train)
````

````python
# now check the accuracy of the model

bg.score(x_test, y_test), bg.score(x_train, y_train)
````

## Use the DecisionTreeClassifier as Estimator

```python

````python
bg = BaggingClassifier(
    estimator= DecisionTreeClassifier(),
    n_estimators=100,
    random_state=42
)

bg.fit(x_train, y_train)
````

````python
# check the accuracy of the model
bg.score(x_test, y_test), bg.score(x_train, y_train)
````

# Random Forest Classifier

````python
from sklearn.ensemble import RandomForestClassifier
````

````python
rf = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)
rf.fit(x_train, y_train)
````

````python
# now check the accuracy of the model

rf.score(x_test, y_test), rf.score(x_train, y_train)
````
