## 📉 Loss Function Explained

A **Loss Function** is a mathematical function that measures the difference between the predicted output of a model ($\hat{y}$) and the actual target value ($y$). It helps the model understand how well or poorly it is performing. The goal of training a model is to minimize this loss.

---

### 🔑 Why is it Important?
- It quantifies the error in predictions.
- It guides the model during training to adjust its parameters (weights and biases) to improve accuracy.

---

### 🛠️ How Does it Work?

1. **Prediction**:
    - The model makes a prediction ($\hat{y}$) based on the input data.

2. **Compare with Actual Value**:
    - The predicted value ($\hat{y}$) is compared with the actual target value ($y$).

3. **Calculate the Error**:
    - The difference between the predicted and actual values is calculated using a specific formula.

4. **Output a Single Value**:
    - The loss function outputs a single value that represents the total error for the entire dataset.

5. **Minimize the Loss**:
    - The model adjusts its parameters (weights and biases) to minimize this loss during training.

---

### 📚 Common Loss Functions

#### 1. **Mean Squared Error (MSE)**  
- Used for regression problems.  
- Measures the average squared difference between predicted and actual values.  

$$
L = \frac{1}{n} \sum_{i=1}^{n} (\hat{y}_i - y_i)^2
$$

- **Steps**:  
  1. Subtract the actual value \( y_i \) from the predicted value \( \hat{y}_i \).  
  2. Square the difference to ensure all errors are positive.  
  3. Take the average of all squared differences.

---

#### 2. **Mean Absolute Error (MAE)**  
- Another loss function for regression.  
- Measures the average absolute difference between predicted and actual values.  

$$
L = \frac{1}{n} \sum_{i=1}^{n} |\hat{y}_i - y_i|
$$

- **Steps**:  
  1. Subtract the actual value \( y_i \) from the predicted value \( \hat{y}_i \).  
  2. Take the absolute value of the difference.  
  3. Take the average of all absolute differences.

---

#### 3. **Cross-Entropy Loss**  
- Used for classification problems.  
- Measures the difference between the predicted probability distribution and the actual distribution.  

$$
L = -\frac{1}{n} \sum_{i=1}^{n} \left[ y_i \log(\hat{y}_i) + (1 - y_i) \log(1 - \hat{y}_i) \right]
$$

- **Steps**:  
  1. For each class, multiply the actual value \( y_i \) by the logarithm of the predicted probability \( \hat{y}_i \).  
  2. Sum these values for all classes.  
  3. Take the negative average of the summed values.


---

#### 4. **Gradient Descent**
- An optimization algorithm used to minimize a loss function.
- Iteratively updates model parameters in the direction that reduces the loss.

$$
\theta := \theta - \alpha \cdot \nabla_\theta J(\theta)
$$

- **Where**:  
  - $( \theta $): Model parameters (e.g., weights)  
  - $( \alpha $): Learning rate (controls step size)  
  - $( J(\theta) $): Loss function  
  - $( \nabla_\theta J(\theta) $): Gradient of the loss function w.r.t. $( \theta $)

---

- **Steps**:
  1. Initialize parameters $( \theta $) randomly or to zero.
  2. Compute the gradient $( \nabla_\theta J(\theta) $) — the direction and rate of steepest ascent of the loss.
  3. Update the parameters by moving in the **opposite direction** of the gradient to minimize the loss.
  4. Repeat until convergence (loss becomes minimal or below a threshold).

---
### 🧮 Example (MAE):
- Suppose we have 3 data points:  
  - Actual values: $y = [3, 5, 2]$  
  - Predicted values: $\hat{y} = [2.5, 5.5, 1.5]$

- Calculate MAE:  
  $$
  L = \frac{1}{3} \left[ |2.5 - 3| + |5.5 - 5| + |1.5 - 2| \right]
  $$
  $$
  L = \frac{1}{3} \left[ 0.5 + 0.5 + 0.5 \right] = 0.5
  $$

---

### 🧮 Example (Cross-Entropy Loss):
- Suppose we have 2 data points for binary classification:  
  - Actual values: $y = [1, 0]$  
  - Predicted probabilities: $\hat{y} = [0.9, 0.2]$

- Calculate Cross-Entropy Loss:  
  $$
  L = -\frac{1}{2} \left[ 1 \cdot \log(0.9) + (1-1) \cdot \log(1-0.9) + 0 \cdot \log(0.2) + (1-0) \cdot \log(1-0.2) \right]
  $$
  $$
  L = -\frac{1}{2} \left[ \log(0.9) + \log(0.8) \right]
  $$
  $$
  L \approx -\frac{1}{2} \left[ -0.1054 + (-0.2231) \right] = 0.1642
  $$

---

### 🧮 Example (Gradient Descent):
- Suppose we have a simple linear model: $y = wx + b$  
- Let's say we want to minimize the loss $J(w) = (wx - y)^2$ for a single data point:  
  - $x = 2$, $y = 4$, initial $w = 1$, learning rate $\alpha = 0.1$

- Compute the gradient of the loss with respect to $w$:
  $$
  \frac{dJ}{dw} = 2(wx - y)x
  $$
  Substitute values:
  $$
  \frac{dJ}{dw} = 2((1 \times 2) - 4) \times 2 = 2(-2) \times 2 = -4 \times 2 = -8
  $$

- Update $w$ using gradient descent:
  $$
  w_{\text{new}} = w - \alpha \cdot \frac{dJ}{dw} = 1 - 0.1 \times (-8) = 1 + 0.8 = 1.8
  $$

- After one step, the new weight $w$ is $1.8$.

---

### 🧮 Example (MSE):
- Suppose we have 3 data points:
  - Actual values: $y = [3, 5, 2]$
  - Predicted values: $\hat{y} = [2.5, 5.5, 1.5]$

- Calculate MSE:
  $$
  L = \frac{1}{3} \left[ (2.5 - 3)^2 + (5.5 - 5)^2 + (1.5 - 2)^2 \right]
  $$
  $$
  L = \frac{1}{3} \left[ 0.25 + 0.25 + 0.25 \right] = 0.25
  $$

---

### 🚀 Summary
- A loss function measures how far off the model's predictions are from the actual values.
- The smaller the loss, the better the model is performing.
- Different loss functions are used depending on the type of problem (regression or classification).
