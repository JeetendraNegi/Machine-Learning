# 💰 Cost Function: Definition, Types, and Working

#### **Definition**
A **cost function** (also called a loss function or error function) is a mathematical formula used in machine learning and statistics to measure how well a model’s predictions match the actual data. It tells us how “wrong” the model is.

#### **Types of Cost Functions**
1. 🧮 **Mean Squared Error (MSE):**
    - Used for regression problems.
    - Calculates the average of the squares of the errors (difference between actual and predicted values).
    - Formula:  
      $$ MSE = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 $$
2. 📏 **Mean Absolute Error (MAE):**
    - Also for regression.
    - Takes the average of the absolute differences between actual and predicted values.
    - Formula:  
      $$ MAE = \frac{1}{n} \sum_{i=1}^{n} |y_i - \hat{y}_i| $$
3. 🔑 **Cross-Entropy Loss (Log Loss):**
    - Used for classification problems.
    - Measures the difference between actual and predicted probability distributions.
    - Formula (binary):  
      $$ -[y \log(\hat{y}) + (1-y) \log(1-\hat{y})] $$
4. ⚖️ **Hinge Loss:**
    - Used for Support Vector Machines (SVM).
    - Focuses on the margin between classes.

#### **Working of Cost Function**
- The cost function calculates a single number that represents how far off the model’s predictions are from the actual results.
- During training, the model tries to **minimize** this cost by adjusting its parameters (like weights in neural networks).
- Lower cost means better performance.
- Optimization algorithms (like Gradient Descent) use the cost function to update the model and improve accuracy.

#### 🆚 Cost Function vs. Loss Function
- **Loss function** usually refers to the error calculated for a single data point.
- **Cost function** is often the average (or sum) of the loss function over the entire dataset.
- In practice, the terms are sometimes used interchangeably, but technically, loss is per sample and cost is overall.

**In simple words:**  
The cost function tells us how bad our model is at making predictions. The goal is to make this number as small as possible by improving the model.
