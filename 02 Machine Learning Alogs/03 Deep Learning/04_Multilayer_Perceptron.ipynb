# 🧠 Multilayer Perceptron (MLP)

A **Multilayer Perceptron (MLP)** is a class of feedforward artificial neural networks (ANN). It consists of at least three layers of nodes: an input layer, one or more hidden layers, and an output layer. Each node (neuron) in one layer is connected to every node in the next layer, making it a fully connected network.

## 🔑 Key Components:
1. **Input Layer**: Accepts the input features of the data.
2. **Hidden Layers**: Perform computations and extract features by applying weights, biases, and activation functions.
3. **Output Layer**: Produces the final output (e.g., classification or regression result).
4. **Weights and Biases**: Parameters that are adjusted during training to minimize the error.
5. **Activation Function**: Introduces non-linearity to the model, enabling it to learn complex patterns.

## ⚙️ Working:
1. **Forward Propagation**:
    - Input data is passed through the network layer by layer.
    - Each neuron computes a weighted sum of its inputs, adds a bias, and applies an activation function.
    - The output of one layer becomes the input to the next layer.

2. **Loss Calculation**:
    - The output of the network is compared to the actual target values using a loss function (e.g., Mean Squared Error, Cross-Entropy).

3. **Backward Propagation**:
    - The error is propagated backward through the network.
    - Gradients of the loss with respect to weights and biases are computed using the chain rule.
    - Weights and biases are updated using optimization algorithms like Gradient Descent.

4. **Training**:
    - The forward and backward propagation steps are repeated for multiple iterations (epochs) until the model converges to an optimal solution.




## 🔄 Forward and Backward Propagation Explained

### 1. **➡️ Forward Propagation**
Forward propagation is the process of passing input data through the network to compute the output. It involves the following steps:

#### **Steps**:
1. **Weighted Sum**:
    Each neuron computes a weighted sum of its inputs:
    $$
    z = \sum_{i=1}^{n} w_i x_i + b
    $$
    where:
    - $w_i$: Weight of the $i$-th input
    - $x_i$: $i$-th input
    - $b$: Bias term
    - $z$: Weighted sum (pre-activation)

2. **Activation Function**:
    The weighted sum is passed through an activation function to introduce non-linearity:
    $$
    a = f(z)
    $$
    where:
    - $f$: Activation function (e.g., ReLU, Sigmoid, Tanh)
    - $a$: Output of the neuron (post-activation)

3. **Layer-by-Layer Computation**:
    The output of one layer becomes the input to the next layer, and this process continues until the output layer produces the final result.

---

### 2. **⬅️ Backward Propagation**
Backward propagation is the process of updating the weights and biases of the network to minimize the error. It uses the chain rule of calculus to compute gradients of the loss function with respect to each parameter.

#### **Steps**:
1. **Loss Function**:
    Compute the error between the predicted output ($\hat{y}$) and the actual target ($y$):
    $$
    L = \text{Loss}(\hat{y}, y)
    $$
    Common loss functions:
    - Mean Squared Error (MSE): $L = \frac{1}{n} \sum_{i=1}^{n} (\hat{y}_i - y_i)^2$
    - Cross-Entropy Loss: $L = -\frac{1}{n} \sum_{i=1}^{n} \left[ y_i \log(\hat{y}_i) + (1-y_i) \log(1-\hat{y}_i) \right]$

2. **Gradient Calculation**:
    Compute the gradient of the loss with respect to each parameter using the chain rule:
    $$
    \frac{\partial L}{\partial w} = \frac{\partial L}{\partial \hat{y}} \cdot \frac{\partial \hat{y}}{\partial z} \cdot \frac{\partial z}{\partial w}
    $$
    Similarly, compute gradients for biases ($b$).

3. **Weight and Bias Update**:
    Update the weights and biases using an optimization algorithm (e.g., Gradient Descent):
    $$
    w = w - \eta \frac{\partial L}{\partial w}
    $$
    $$
    b = b - \eta \frac{\partial L}{\partial b}
    $$
    where:
    - $\eta$: Learning rate (controls the step size of updates)

4. **Repeat**:
    The forward and backward propagation steps are repeated for multiple iterations (epochs) until the loss converges to a minimum.

---

### Summary
- **➡️ Forward Propagation** computes the output of the network.
- **⬅️ Backward Propagation** adjusts the weights and biases to minimize the error.
- Together, they enable the network to learn from data and improve its predictions over time.


## 📉 Loss Function Explained

A **Loss Function** is a mathematical function that measures the difference between the predicted output of a model ($\hat{y}$) and the actual target value ($y$). It helps the model understand how well or poorly it is performing. The goal of training a model is to minimize this loss.

---

### 🔑 Why is it Important?
- It quantifies the error in predictions.
- It guides the model during training to adjust its parameters (weights and biases) to improve accuracy.

---

### 🛠️ How Does it Work?

1. **Prediction**:
    - The model makes a prediction ($\hat{y}$) based on the input data.

2. **Compare with Actual Value**:
    - The predicted value ($\hat{y}$) is compared with the actual target value ($y$).

3. **Calculate the Error**:
    - The difference between the predicted and actual values is calculated using a specific formula.

4. **Output a Single Value**:
    - The loss function outputs a single value that represents the total error for the entire dataset.

5. **Minimize the Loss**:
    - The model adjusts its parameters (weights and biases) to minimize this loss during training.

---

### 📚 Common Loss Functions

#### 1. **Mean Squared Error (MSE)**  
- Used for regression problems.  
- Measures the average squared difference between predicted and actual values.  

$$
L = \frac{1}{n} \sum_{i=1}^{n} (\hat{y}_i - y_i)^2
$$

- **Steps**:  
  1. Subtract the actual value \( y_i \) from the predicted value \( \hat{y}_i \).  
  2. Square the difference to ensure all errors are positive.  
  3. Take the average of all squared differences.

---

#### 2. **Mean Absolute Error (MAE)**  
- Another loss function for regression.  
- Measures the average absolute difference between predicted and actual values.  

$$
L = \frac{1}{n} \sum_{i=1}^{n} |\hat{y}_i - y_i|
$$

- **Steps**:  
  1. Subtract the actual value \( y_i \) from the predicted value \( \hat{y}_i \).  
  2. Take the absolute value of the difference.  
  3. Take the average of all absolute differences.

---

#### 3. **Cross-Entropy Loss**  
- Used for classification problems.  
- Measures the difference between the predicted probability distribution and the actual distribution.  

$$
L = -\frac{1}{n} \sum_{i=1}^{n} \left[ y_i \log(\hat{y}_i) + (1 - y_i) \log(1 - \hat{y}_i) \right]
$$

- **Steps**:  
  1. For each class, multiply the actual value \( y_i \) by the logarithm of the predicted probability \( \hat{y}_i \).  
  2. Sum these values for all classes.  
  3. Take the negative average of the summed values.


---

### 🧮 Example (MSE):
- Suppose we have 3 data points:
  - Actual values: $y = [3, 5, 2]$
  - Predicted values: $\hat{y} = [2.5, 5.5, 1.5]$

- Calculate MSE:
  $$
  L = \frac{1}{3} \left[ (2.5 - 3)^2 + (5.5 - 5)^2 + (1.5 - 2)^2 \right]
  $$
  $$
  L = \frac{1}{3} \left[ 0.25 + 0.25 + 0.25 \right] = 0.25
  $$

---

### 🚀 Summary
- A loss function measures how far off the model's predictions are from the actual values.
- The smaller the loss, the better the model is performing.
- Different loss functions are used depending on the type of problem (regression or classification).
