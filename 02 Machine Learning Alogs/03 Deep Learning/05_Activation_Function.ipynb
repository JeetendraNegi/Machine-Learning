# 🔌 Activation Function in Deep Learning
**Definition:**  
An activation function is a mathematical function used in neural networks to decide whether a neuron should be activated or not. It introduces non-linearity into the model, allowing the network to learn complex patterns.

---

#### 🗂️ Types of Activation Functions

1. **Linear Activation Function**
    - Output is proportional to the input.
    - Not commonly used in deep networks because it cannot capture complex patterns.

2. **Non-linear Activation Functions:**
    - **Sigmoid:**  
      Output ranges between 0 and 1. Good for binary classification problems, especially in the output layer.
      $$
      \sigma(x) = \frac{1}{1 + e^{-x}}
      $$
    - **Tanh (Hyperbolic Tangent):**  
      Output ranges between -1 and 1. Zero-centered, often used in hidden layers of Recurrent Neural Networks (RNNs).
      $$
      \tanh(x) = \frac{e^x - e^{-x}}{e^x + e^{-x}}
      $$
    - **ReLU (Rectified Linear Unit):**  
      Output is 0 if input is negative, otherwise output is input itself. Commonly used in hidden layers of deep neural networks due to its simplicity and effectiveness.
      $$
      f(x) = \max(0, x)
      $$
    - **Leaky ReLU:**  
      Similar to ReLU but allows a small, non-zero gradient when input is negative. Helps prevent the "dying ReLU" problem.
    - **Softmax:**  
      Used in the output layer for multi-class classification problems (categorical outputs). Converts a vector of values into probabilities that sum to 1.
      $$
      \text{Softmax}(x_i) = \frac{e^{x_i}}{\sum_{j} e^{x_j}}
      $$

---

#### ⚙️ How Activation Functions Work

- Each neuron in a neural network receives input, applies a weighted sum, and then passes the result through an activation function.
- The activation function decides the output of the neuron, introducing non-linearity.
- Without activation functions, the neural network would behave like a simple linear model, unable to solve complex tasks.

---

#### 💡 When to Use Which Activation Function?

- **Sigmoid:** Use in the output layer for binary classification tasks.
- **Tanh:** Preferred in hidden layers of RNNs due to its zero-centered output.
- **ReLU:** Default choice for hidden layers in most deep learning models (CNNs, DNNs).
- **Leaky ReLU:** Use when you face the "dying ReLU" problem (many neurons outputting zero).
- **Softmax:** Use in the output layer for multi-class (categorical) classification problems.

---

**In summary:**  
Activation functions help neural networks learn and represent complex data by introducing non-linearity, making them essential for deep learning models. Choosing the right activation function depends on the type of problem and the network architecture.
