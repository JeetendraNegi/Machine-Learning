## 📌 **Chi-Square Test**  

The **Chi-Square Test** is a statistical test used to determine whether there is a significant association between two categorical variables. It compares observed data with expected data under the assumption of independence.

### **Formula for Chi-Square Test**  

$$
\chi^2 = \sum \frac{(O - E)^2}{E}
$$

Where:  
- \( O \) = Observed frequency  
- \( E \) = Expected frequency  

### **Types of Chi-Square Tests**  
1. **Chi-Square Test for Independence** – Used to determine if two categorical variables are related.  
2. **Chi-Square Goodness-of-Fit Test** – Used to determine if a sample distribution fits a population distribution.  


```python
#import the required packages
import numpy as np
import scipy.stats as st
import matplotlib.pyplot as plt
```

## Example 1:
A fair die is rolled 120 times and the following results are obtained:

Face 1: 22 times

Face 2: 17 times

Face 3: 20 times

Face 4: 26 times

Face 5: 22 times

Face 6: 13 times

Test at a 5% level of significance whether the die is fair.

# solution:
we will assume data is normally distributed and decide the H0 and Ha

H0 = dice is fare

Ha = dice is not fare

```python
# preparing the data
observation_values =np.array([22, 17, 20, 26, 22, 13])
expected_values= np.array([20] * 6) # this will give a list with 6 length with all 20
alpha = 0.05
df = 6 - 1 # n - 1, we have total 6 observations
```

```python
# get the chi square table value (Observation)
chi_square_table = st.chi2.ppf(1 - alpha, df)
chi_square_table
```

```python
#calculate the chi square value using formula
chi_square_cal = np.sum(np.square(observation_values - expected_values)/(expected_values))
chi_square_cal
```

```python
if chi_square_cal > chi_square_table:
    print('Ha is right. means dice is not fare')
else:
    print('H0 is right. means dice is fare')
```

```python
# plot the graph
x = np.linspace(0, 20, 500)
y = st.chi2.ppf(x, df)

plt.figure(figsize=(8, 5))
plt.plot(x, y, label=f'chi-square distribution (df={df})', color='red')

#mark the chi square table value
plt.axvline(chi_square_table, color='red', linestyle='--', label= f'chi-square-table value ={chi_square_table}')
plt.fill_between(x, y, where=(x >= chi_square_table), color='red', alpha=0.05)

# mark the chi square calculated
plt.axvline(chi_square_cal, color='blue', linestyle='-', label=f'chi-square-calculated value = {chi_square_cal}')

#plot

plt.xlabel('chi-square Value')
plt.ylabel('Probability Density')
plt.title('Chi-square test for fairness of die')
plt.legend()
plt.grid()
plt.show()
```

# **Example 2:**

A study was conducted to investigate whether there is a relationship between gender and the preferred genre of music. A sample of 235 people was selected, and the data collected is shown below. Test at a 5% level of significance whether there is a significant association between gender and music preference.

|        | **Pop** | **Hip Hop** | **Classical** | **Rock** |
|--------|--------|------------|--------------|---------|
| **Male**   | 40     | 45         | 25           | 10      |
| **Female** | 35     | 30         | 20           | 30      |



```python
# Data Table
# Rows: Gender (Male, Female)
# Columns: Genre (Pop, Hip Hop, Classical, Rock)
data = np.array([[40, 45, 25, 10],  # Male
                 [35, 30, 20, 30]]) # Female

# Performing the Chi-Square Test
chi2_stat, p_value, dof, expected = st.chi2_contingency(data)

# Print results
print(f"Chi-Square Statistic: {chi2_stat}")
print(f"P-Value: {p_value}")
print(f"Degrees of Freedom: {dof}")
print("Expected Frequencies:")
print(expected)

# Conclusion
alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis: There is a significant association between gender and music preference.")
else:
    print("Fail to reject the null hypothesis: No significant association between gender and music preference.")
```
