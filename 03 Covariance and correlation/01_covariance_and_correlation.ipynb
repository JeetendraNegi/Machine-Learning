# 📌 Covariance vs. Correlation

## 🔹 Covariance Definition
Covariance measures the direction of the relationship between two random variables. It tells us whether two variables increase or decrease together but does not standardize the relationship.

# 🔹 Correlation Definition
Correlation measures both the strength and direction of a linear relationship between two variables. It is the standardized version of covariance and ranges between -1 and 1.

```python
# importing the required packages

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# loading the data

dataset = pd.read_excel("Financial Sample.xlsx")
dataset.head(3)
```

```python
# getting the data types of the dataset
dataset.info
```

```python
# getting the covariance and correlation
data_cov = dataset.select_dtypes(['float64', 'int64']).cov()
data_cov

data_corr = dataset.select_dtypes(['float64', 'int64']).corr()
data_corr
```

```python
# plot the data in the graph

sns.heatmap(data_corr)
plt.show()
```

```python
# plot the covariance 
sns.heatmap(data_cov)
plt.show()
```
