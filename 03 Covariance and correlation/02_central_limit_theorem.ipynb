# 📌 Central Limit Theorem (CLT) – Simple Definition for Everyone
The Central Limit Theorem (CLT) says that if you take a lot of small random groups (samples) from any kind of data, and calculate their average, those averages will start to form a bell-shaped curve (normal distribution) as the sample size gets bigger.

## 🔹 Think of it Like This:
Imagine you have a jar full of different-sized marbles.

If you randomly pick 5 marbles and find their average size, the result might be all over the place.

But if you keep taking more and more samples (like 30 or more), and find the average size of each sample, those averages will start forming a smooth, bell-shaped pattern.

💡 Even if the original data is messy, the averages will always move toward a normal distribution!

This is why CLT is important in statistics—it helps make predictions, analyze data, and apply probability even when the original data is random or uneven! 😊

```python
# import the required 
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# generating the data
pop_data = [np.random.randint(10, 100) for i in range(10000)]
pop_table = pd.DataFrame({"Pop_Data": pop_data})
pop_table.head(10)
```

```python
#plot the data

sns.kdeplot(x="Pop_Data", data= pop_table)
plt.show()
```

```python
# now we taking the 50 sample of this data that should not greater then 10% of total data

sample_mean = []
for no_sample in range(100):
    sample_data = []
    for data in range(900): # getting only 500 data from data set
        sample_data.append(np.random.choice(pop_data))
    sample_mean.append(np.mean(sample_data))
```

```python
sample_M = pd.DataFrame({'Sample_Mean': sample_mean})

#display the plot

sns.kdeplot(x = 'Sample_Mean', data = sample_M)
plt.show()
```

```python
# as per central limit theorem mean of the total data and mean of the mean of the sample data will be almost same

mean_of_the_data = np.mean(pop_data)
mean_of_mean_of_sample_data = np.mean(sample_mean)

mean_of_the_data, mean_of_mean_of_sample_data
```
