# 📌 Hypothesis Testing – Definition & Types  

## 🔹 Definition of Hypothesis Testing  
Hypothesis testing is a **statistical method** used to **make decisions** or **draw conclusions** about a population based on a **sample of data**. It helps determine whether there is enough **evidence** to support a claim or if we should reject it.  

✅ It works by **comparing data** against an **assumption (hypothesis)** and checking if the results are **due to chance or a real effect**.  

---

## 🔹 Types of Hypotheses in Testing  

### 1️⃣ **Null Hypothesis (H₀)**  
   - This is the **default assumption** that **there is no effect or difference**.  
   - **Example:** "A new medicine has **no effect** on blood pressure."  
   - We try to **disprove** this hypothesis.  

### 2️⃣ **Alternative Hypothesis (H₁ or Ha)**  
   - This states that **there is an effect or difference**.  
   - **Example:** "A new medicine **lowers blood pressure**."  
   - If there is **strong evidence**, we accept this hypothesis.  
 

---

## 📌 **Key Takeaways**  
✔ **Hypothesis testing helps us make decisions** about a population based on data.  
✔ The **null hypothesis (H₀)** assumes **no effect**, and the **alternative hypothesis (H₁)** assumes an **effect**.  
✔ We use **one-tailed** tests when we expect a specific direction and **two-tailed** tests when we are checking for any change.
