# 📌 Steps of Hypothesis Testing  

Hypothesis testing follows a structured process to determine whether a claim about a population is supported by sample data.  

---

## 🔹 **Step 1: Define the Hypotheses**  
- **Null Hypothesis (H₀):** The assumption that there is **no effect or difference**.  
- **Alternative Hypothesis (H₁ or Ha):** The assumption that there **is** an effect or difference.  

🔹 **Example:**  
H₀: A new drug **does not** lower blood pressure.  
H₁: A new drug **lowers** blood pressure.  

---

## 🔹 **Step 2: Choose the Significance Level (α)**  
- The **significance level (α)** represents the probability of rejecting H₀ when it is actually true.  
- Common values: **0.05 (5%)** or **0.01 (1%)**.  

✅ **Example:** If α = 0.05, there is a **5% chance** of wrongly rejecting H₀.  

---

## 🔹 **Step 3: Select the Appropriate Test**  
Choose a statistical test based on:  
- **Type of data** (categorical or numerical).  
- **Sample size** (small or large).  
- **Distribution** (normal or non-normal).  

✔ **Common Tests:**  
- **Z-Test / T-Test** (Comparing means)  
- **Chi-Square Test** (Categorical data)  
- **ANOVA** (Comparing multiple groups)  

---

## 🔹 **Step 4: Calculate the Test Statistic & P-Value**  
- Compute the **test statistic** (Z-score, T-score, etc.).  
- Find the **p-value**, which tells how likely it is to get the observed data if H₀ is true.  

✅ **If p-value ≤ α**, reject the null hypothesis (**H₀**).  
✅ **If p-value > α**, fail to reject H₀ (not enough evidence).  

---

## 🔹 **Step 5: Make a Decision**  
- If **p-value ≤ α** → Reject H₀ → The result is **statistically significant**.  
- If **p-value > α** → Fail to reject H₀ → Not enough evidence to support H₁.  

🔹 **Example Decision:**  
- "Since p-value **0.03 < 0.05**, we reject H₀. The new drug significantly lowers blood pressure."  

---

## 📌 **Summary of Hypothesis Testing Steps**  
1️⃣ **Define the null and alternative hypotheses (H₀ & H₁).**  
2️⃣ **Choose the significance level (α), usually 0.05 or 0.01.**  
3️⃣ **Select the appropriate statistical test.**  
4️⃣ **Calculate the test statistic and p-value.**  
5️⃣ **Make a decision (Reject or Fail to Reject H₀).**  

🎯 **This process helps in making data-driven decisions with confidence!** 🚀
