# 📌 T-Test – Definition & Formula  

## 🔹 **Definition of T-Test**  
The **T-Test** is a statistical test used to compare the means of one or two groups and determine whether the differences between them are statistically significant.  

✔ Used when the **sample size is small (n < 30)**.  
✔ Used when the **population standard deviation (σ) is unknown**.  
✔ Assumes data follows a **normal distribution** (or approximately normal).  

---

## 🔹 **Types of T-Test**  
1️⃣ **One-Sample T-Test** – Compares the sample mean to a known population mean.  
2️⃣ **Two-Sample (Independent) T-Test** – Compares the means of two independent groups.  
3️⃣ **Paired T-Test** – Compares the means of the same group before and after treatment.  

---

## 🔹 **Formula for T-Test**  

### **1️⃣ One-Sample T-Test**  
When testing whether a sample mean **(𝑥̄)** is different from the population mean **(μ)**:  

$$
t = \frac{\bar{x} - \mu}{\frac{s}{\sqrt{n}}}
$$  

🔹 Where:  
- $( t $) = t-score (test statistic)  
- $( \bar{x} $) = Sample mean  
- $( \mu $) = Population mean  
- $( s $) = Sample standard deviation  
- $( n $) = Sample size  

---

### **2️⃣ Two-Sample (Independent) T-Test**  
When comparing two sample means **(𝑥̄₁ and 𝑥̄₂)** from two independent groups:  

$$
t = \frac{(\bar{x}_1 - \bar{x}_2)}{\sqrt{\frac{s_1^2}{n_1} + \frac{s_2^2}{n_2}}}
$$  

🔹 Where:  
- $( \bar{x}_1, \bar{x}_2 $) = Sample means of groups 1 and 2  
- $( s_1, s_2 $) = Sample standard deviations of groups 1 and 2  
- $( n_1, n_2 $) = Sample sizes of groups 1 and 2  

---

### **3️⃣ Paired T-Test**  
When comparing the means of the same group before and after a treatment:  

$$
t = \frac{\bar{d}}{\frac{s_d}{\sqrt{n}}}
$$  

🔹 Where:  
- $( \bar{d} $) = Mean of the differences between paired samples  
- $( s_d $) = Standard deviation of the differences  
- $( n $) = Number of paired observations  

---

## 📌 **When to Use a T-Test?**  
✔ **One-Sample T-Test** → Comparing a sample mean to a known population mean.  
✔ **Two-Sample (Independent) T-Test** → Comparing two independent groups (e.g., males vs. females).  
✔ **Paired T-Test** → Comparing measurements before and after a treatment (e.g., pre-test vs. post-test).  

---

## 🎯 **Conclusion**  
The T-Test helps determine if the difference between means is **statistically significant**. It is used in hypothesis testing when **sample sizes are small** and the **population standard deviation is unknown**. 🚀


# Example 1: 
a manufacturer claims that the average weight of a bag of a potato chips is 150 grams. a sample of 25 bags is taken, and the average weight is found to be 148 grams, with a standard deviation of 5 grams. test the manufacturer's claim using a one-tailed t-test with a significance level of 0.05.

```python
#import the required packages
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import scipy.stats as st
```

```python
# preparing the data. these are given data in the example

sample_std = 5
no_sample = 25
alpha = 0.05
mean_pop = 150
mean_sample = 148
df = 24 # degree of freedom is (n - 1)
```

```python
# getting the t table value
t_table = st.t.ppf(1 - alpha, df)
t_table
```

```python
# calculating the t values
t_cal = (148 - 150)/(sample_std/np.sqrt(no_sample))
t_cal
```

```python
if t_cal > t_table:
    print('Ha is right')
else:
    print('H0 is right')
```

# Example 2:
A company wants to test whether there is a difference in productivity between two teams. They randomly select 20 employees from each team and record their productivity scores. The mean productivity score from team A is 80 with a std of 5, while the mean productivity score of team B is 75 with a std of 6. Test at a 5% level of significance whether there is a difference of productivity between the tow teams.

```python
# data preparation
teamA_mean = 80
teamB_mean = 75
no_sample = 20
teamA_std = 5
teamB_std = 6
alpha = 0.05
df = (20 -1) + (20 - 1)
```

```python
# calculating the t table value

t_table = st.t.pdf(1- alpha, df)
t_table
```

```python
# calculating the t value

t_cal = (teamA_mean - teamB_mean)/np.sqrt((teamA_std**2/no_sample)+(teamB_std**2/no_sample))
t_cal
```

```python
if t_cal == 0:
    print('H0 is right,means there is no difference')
else:
    print('Ha is right,means there is some difference')
```

# Example 3:
A company wants to test whether a new training program improves the typing speed of its employees. The typing speed of 20 employees was recorded before and after the training program. The data is given below. Test at a 5% level of significance whether the training program has an effect on the typing speed of the employees.

Before: 50, 60, 45, 65, 55, 70, 40, 75, 80, 65, 70, 60, 50, 55, 45, 75, 60, 50, 65

After:  60, 70, 55, 75, 65, 80, 50, 85, 90, 70, 75, 65, 55, 60, 50, 80, 65, 55, 75

```python
# preparing the data
no_sample = 20
alpha = 0.05/2

data_old = np.array([50, 60, 45, 65, 55, 70, 40, 75, 80, 65, 70, 60, 50, 55, 45, 75, 60, 50, 65])
data_new = np.array([60, 70, 55, 75, 65, 80, 50, 85, 90, 70, 75, 65, 55, 60, 50, 80, 65, 55, 75])

#getting the mean
data_old_mean = np.mean(data_old)
data_new_mean = np.mean(data_new)

# getting the std
data_old_std = data_old.std()
data_new_std = data_new.std()

df = 20 - 1
```

```python
# getting t table value

t_table = st.t.ppf(1 - alpha, df)
t_table
```

```python
# getting t calculating
t_cal = (data_new_mean - data_old_mean)/ np.sqrt((data_new_std**2/no_sample) + (data_old_std**2/no_sample))
t_cal
```

```python
if t_cal > t_table:
    print('Ha is right, training program has an effect on typing speed.')
else:
    print('H0 is right, training program has not any effect on typing speed.')
```
