mean of the data is

sum of the data/ total number of the data

median = center data (if number of records are odd, if even the sum the middle data/2)

```python
import numpy as np
import pandas as pd
from scipy import stats
```

```python
ar = np.array([1,2,3,53,5,45,4,32,12, 45])

sum_arr = ar.sum()
print(f'sum of the array {sum_arr}')

length = len(ar)
print(f'Length of the array: {length}')

mean = sum_arr/length
print(f'mean of the array is {mean}')

ar_median = np.median(ar)
print(f'median of the array is {ar_median}') 

# Correct mode calculation using scipy.stats.mode
ar_mode = stats.mode(ar, keepdims=True)[0][0]
print(f'Mode of the array: {ar_mode}')

```
