## Percentile and Quartile

```python
#import the required library

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# import the dataset 

dataset = pd.read_excel('Financial Sample.xlsx')
dataset.head(3)
```

```python
dataset.describe()
```

```python
# getting the Percentile of the Gross Sales

np.percentile(dataset['Gross Sales'], 90), np.percentile(dataset['Gross Sales'], 0)
```

```python
#display the data

sns.boxenplot(x= "Gross Sales", data = dataset)
plt.show()
```
