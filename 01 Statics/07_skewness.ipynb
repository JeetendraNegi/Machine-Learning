## 📌 Skewness Definition & Formula

The **Skewness** of a dataset measures the **asymmetry** of its distribution. It tells us whether the data is skewed **left (negative skew)**, **right (positive skew)**, or **symmetrically distributed (zero skewness).**

### 🔹 **Formula for Skewness**
$$
S = \frac{\sum (X_i - \bar{X})^3}{n \cdot s^3}
$$

Where:
- \( S \) = Skewness  
- \( X_i \) = Each data point  
- \( \bar{X} \) = Mean of the dataset  
- \( n \) = Total number of data points  
- \( s \) = Standard deviation  
- \( (X_i - \bar{X})^3 \) = Cubed deviation from the mean  

---

### 📌 **Interpreting Skewness:**
- **\( S = 0 \)** → **Symmetrical distribution (normal distribution)**
- **\( S > 0 \)** → **Positively skewed (Right-skewed, long right tail)**
- **\( S < 0 \)** → **Negatively skewed (Left-skewed, long left tail)**

---

### 🔹 **Alternative Formula (Pearson’s Second Skewness Coefficient)**
$$
Sk = \frac{3(\bar{X} - \text{Median})}{s}
$$
Where:
- \( Sk \) = Pearson’s skewness coefficient  
- \( \bar{X} \) = Mean  
- **Median** = Middle value of the dataset  
- \( s \) = Standard deviation  

---

```python
# Testing with random data

data = np.random.normal(20, 100, 100)
```

```python
# check the skewness

df = pd.DataFrame({"x": data}) # this is create a table with data with a new column 'x' and value as data
df["x"].skew()
```

```python
# checking the mean median mod

df['x'].mean(), df['x'].median(), df['x'].mode()[0]
```

```python
sns.histplot(x="x", data = df)
plt.show()
```
