### The MAD of a dataset is the average distance between each data point and the mean. it gives us an idea about the variability in a dataset

## 📌 Mean Absolute Deviation (MAD) Formula

The **Mean Absolute Deviation (MAD)** measures the **average absolute deviation** of each data point from the mean.

$$
MAD = \frac{\sum |X_i - \mu|}{N}
$$

Where:
- \( X_i \) = Each data point  
- \( \mu \) = Mean of the dataset  
- \( N \) = Total number of data points  
- \( |X_i - \mu| \) = Absolute deviation from the mean  

​


```python
#import the required library

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# preparing the data for test

sec_a = np.array([75, 65, 73, 68, 72, 76])

sec_b = np.array([90, 47, 43, 96, 93, 51])

# preparing the column for number of data

no = np.array([1,2,3,4,5,6])

```

```python
# getting the mean of any one because both mean are same

mean = np.mean(sec_b)
```

```python
#getting the mean absolute deviation
# get the Xi - U   --> sec_a - mean
# get the mod of this | Xi - U| using abs (absolute value function) it will return |+-2| = 2
# then get the sum of the | Xi - U | using sum function
# then divide it by total number of element in the sec_a or sec_b

step1 = sec_a - mean
step2 = np.abs(step1)
step3 = np.sum(step2)
step4 = step3/len(sec_a) # this is mean absolute deviation of sec_a

# in a single step

mad_a = np.sum(np.abs(sec_a - mean))/len(sec_a)
mad_b = np.sum(np.abs(sec_b - mean))/len(sec_b)

mad_a, mad_b
```

```python
# plot the data in the graf. [mean]*6 will create a list like [70, 70, 70, 70, 70, 70] mean with 6 time.

plt.figure(figsize=(10, 3))
plt.scatter(sec_a, no, label='sec A')
plt.scatter(sec_b, no, label= 'sec B')
plt.plot([mean]*6, no, c = 'blue', label='mean')
plt.legend()
plt.show()
```
