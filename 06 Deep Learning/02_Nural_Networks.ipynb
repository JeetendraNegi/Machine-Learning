# 🤖 Neural Networks: A Simple Explanation

## 🧠 What is a Neural Network?

A **neural network** is a computer system inspired by the human brain. It is designed to recognize patterns and solve problems by learning from data.

---

## ⚙️ How Does a Neural Network Work?

1. **Input Layer**  
    - Takes in the raw data (like images, numbers, or text).
2. **Hidden Layers**  
    - One or more layers that process the input using mathematical operations.
    - Each layer consists of "neurons" (nodes) that transform the data.
3. **Output Layer**  
    - Produces the final result (like a prediction or classification).

---

## 📝 Step-by-Step Working

1. **Data Input:**  
    - Data is fed into the input layer.
2. **Weighted Sum:**  
    - Each neuron multiplies the input by a weight and adds a bias.
3. **Activation Function:**  
    - The result passes through an activation function (like ReLU or Sigmoid) to introduce non-linearity.
4. **Forward Propagation:**  
    - The output from one layer becomes the input for the next layer.
5. **Output Generation:**  
    - The final layer produces the output.
6. **Loss Calculation:**  
    - The difference between the predicted output and the actual value is calculated (loss).
7. **Backpropagation:**  
    - The network adjusts the weights and biases to reduce the loss using optimization algorithms (like Gradient Descent).
8. **Repeat:**  
    - Steps 1–7 are repeated for many cycles (epochs) until the network learns well.

---

## 🧩 Types of Neural Networks

1. **Feedforward Neural Network (FNN):**  
    - Data moves in one direction, from input to output.
    - Used for simple tasks like classification.

2. **Convolutional Neural Network (CNN):**  
    - Specializes in processing images.
    - Uses convolutional layers to detect patterns like edges and shapes.

3. **Recurrent Neural Network (RNN):**  
    - Designed for sequential data (like text or time series).
    - Has loops to remember previous information.

4. **Generative Adversarial Network (GAN):**  
    - Consists of two networks (generator and discriminator) that compete to create realistic data.

5. **Radial Basis Function Network (RBFN):**  
    - Uses radial basis functions as activation functions.
    - Good for pattern recognition.

| Type      | Best For           | Key Feature                |

## 🏁 In Short

- Neural networks learn from data by adjusting weights and biases.
- They can solve complex problems like image recognition, language translation, and more.
- Different types are used for different tasks.
