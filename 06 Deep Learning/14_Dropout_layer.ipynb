## 🧩 Dropout Layer in Deep Learning

**📝 Definition:**  
A dropout layer is a regularization technique used in deep learning to prevent overfitting. It works by randomly "dropping out" (setting to zero) a fraction of the neurons during training.

**🎯 Purpose:**  
- To reduce overfitting by making the model less sensitive to specific neurons.
- To encourage the network to learn more robust and general features.

**⚙️ How It Works:**  
- During training, each neuron has a probability (e.g., 0.5) of being temporarily ignored.
- The dropped neurons do not contribute to the forward pass or backpropagation.
- During testing, all neurons are used, but their outputs are scaled down to account for the dropout during training.

**💡 In Simple Words:**  
Dropout randomly turns off some neurons while training so the model doesn't rely too much on any one part. This helps the model perform better on new, unseen data.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# import the dataset

dataset = pd.read_csv('Churn_Modelling.csv')
dataset.head(3)
```

```python
# check the null values
dataset.isnull().sum()
```

```python
# remove the unused features(columns)
dataset = dataset.drop(columns=['RowNumber', 'CustomerId', 'Surname', 'Geography', 'Gender'], axis=1)
dataset.head(3)
```

```python
# input and output features

input_data = dataset.iloc[:, :-1]
output_data = dataset.iloc[:, -1]
```

```python
# scale the input data

from sklearn.preprocessing import StandardScaler

ss = StandardScaler()
```

```python
input_data = pd.DataFrame(ss.fit_transform(input_data), columns=input_data.columns)
input_data.head(3)
```

```python
# train test split the data

from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(input_data, output_data, test_size=0.2, random_state=42)
```

```python
import tensorflow as tf
from keras.layers import Dense, BatchNormalization, Dropout
from keras.models import Sequential
from keras.regularizers import L2
from keras.callbacks import EarlyStopping

from sklearn.metrics import accuracy_score
```

```python
ann = Sequential()
```

```python
# create the hidden layers, taking the 6 layers because input_data.shape has 8 features so we take less then 8
ann.add(Dense(units=6, input_dim=8, activation='relu', kernel_regularizer=L2(l2=0.01)))
ann.add(BatchNormalization())
ann.add(Dropout(0.2))  # dropout layer
ann.add(Dense(units=4, activation='relu'))  # you could use regularization here too
ann.add(BatchNormalization())
ann.add(Dropout(0.2))  # dropout layer
ann.add(Dense(units=2, activation='relu'))  # you could use regularization here too
ann.add(BatchNormalization())
ann.add(Dropout(0.2))  # dropout layer
ann.add(Dense(units=1, activation='sigmoid'))  # output layer
```

```python
# compile the model
ann.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python

ann.fit(x_train, y_train, batch_size=100, epochs=50, validation_data=(x_test, y_test), callbacks= EarlyStopping()) 
```

```python
# check the accuracy of the model

accuracy_score(y_train, ann.predict(x_train) > 0.5), accuracy_score(y_test, ann.predict(x_test) > 0.5)
```
