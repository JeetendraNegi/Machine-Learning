# 🛒 Association Rule Learning

## 📘 Definition

**Association Rule Learning** is a rule-based machine learning technique used to **discover relationships between variables in large datasets**.  
It is most commonly used in **market basket analysis**, where we want to find which products are frequently bought together.

For example:  
"If a customer buys bread and butter, they are likely to buy milk."

---

## 📊 Key Terms and Metrics

### 1. **Itemset**
- A collection of one or more items.
- Example: {Bread, Butter, Milk}

### 2. **Support**
- How frequently an itemset appears in the dataset.
- It tells us the **popularity** of an item or itemset.

$[
\text{Support}(A) = \frac{\text{Number of transactions containing A}}{\text{Total number of transactions}}
$]

---

### 3. **Confidence**
- The likelihood that item **B** is bought when item **A** is bought.

$[
\text{Confidence}(A \rightarrow B) = \frac{\text{Support}(A \cup B)}{\text{Support}(A)}
$]

- Example: If 100 people bought bread, and 60 of them also bought butter, confidence of `Bread → Butter` is 60%.

---

### 4. **Lift**
- Tells us how much more likely **B** is purchased when **A** is purchased, compared to B being purchased independently.

$[
\text{Lift}(A \rightarrow B) = \frac{\text{Confidence}(A \rightarrow B)}{\text{Support}(B)}
$]

- Lift > 1: Positive association (they occur together more than random chance)
- Lift = 1: No association
- Lift < 1: Negative association

---

## 🔁 Working Step-by-Step (Apriori Algorithm)

### Step 1: Set thresholds
- Decide **minimum support** and **confidence**.

### Step 2: Generate Frequent Itemsets
- Count item combinations that meet **minimum support**.

### Step 3: Generate Association Rules
- For each frequent itemset, create rules like `A → B`, `A, C → D`, etc.

### Step 4: Filter Rules
- Keep rules that meet the **minimum confidence** threshold.

### Step 5: Evaluate with Lift
- Calculate **lift** to find how strong the rule is beyond chance.

---

## ✅ Example

| Transaction ID | Items                     |
|----------------|---------------------------|
| 1              | Bread, Butter             |
| 2              | Bread, Butter, Milk       |
| 3              | Milk                      |
| 4              | Bread, Milk               |

From this data:

- Support(Bread) = 3/4 = 0.75  
- Support(Milk) = 3/4 = 0.75  
- Support(Bread ∩ Milk) = 2/4 = 0.5  
- Confidence(Bread → Milk) = 0.5 / 0.75 = 0.67  
- Lift(Bread → Milk) = 0.67 / 0.75 ≈ 0.89

This tells us that buying bread actually **reduces** the likelihood of milk slightly (Lift < 1).

---

## 📚 Common Algorithms

### 1. **Apriori**
- Generates all frequent itemsets using a bottom-up approach.
- Removes combinations that don't meet the support threshold.$
### 2. **ECLAT (Equivalence Class Clustering and bottom-up Lattice Traversal)**
- Uses a depth-first approach for mining frequent itemsets.
- Works faster for dense datasets.

### 3. **FP-Growth (Frequent Pattern Growth)**
- Uses a compact tree (FP-tree) to avoid generating all possible itemsets.
- More efficient than Apriori for large datasets.

---

## 📌 Summary

| Metric     | Meaning                          |
|------------|-----------------------------------|
| Support    | How often the itemset appears     |
| Confidence | How often rule is true            |
| Lift       | How strong the rule is (beyond chance) |

| Algorithm   | Approach         | Best For                |
|-------------|------------------|--------------------------|
| Apriori     | Bottom-up search | Simpler, smaller datasets |
| ECLAT       | Depth-first search | Dense data               |
| FP-Growth   | Tree-based       | Large-scale datasets     |

---

## 🧠 Real-Life Use Cases

- Market basket analysis (e.g., retail, e-commerce)
- Recommender systems (e.g., "People who bought X also bought Y")
- Web usage mining
- Bioinformatics (gene association)
