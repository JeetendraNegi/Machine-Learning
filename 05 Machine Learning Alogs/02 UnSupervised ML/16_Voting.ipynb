### 📚 Definitions and Examples

#### 1. **Max Voting** 🗳️
Max Voting is a technique used in ensemble methods where multiple models make predictions, and the final prediction is determined by the majority vote. The class that receives the highest number of votes is selected as the final output.

**Example:**
Three models predict the class of an image:
- Model 1: 🐱 Cat
- Model 2: 🐶 Dog
- Model 3: 🐱 Cat

**Final Prediction:** 🐱 Cat (as it has the majority votes).

---

#### 2. **Averaging** 📊
In Averaging, the predictions from multiple models are averaged (for regression tasks) or the probabilities of each class are averaged (for classification tasks). The final prediction is based on the highest average probability.

**Example:**
Three models predict probabilities for two classes (🐱 Cat and 🐶 Dog):
- Model 1: [0.7, 0.3]
- Model 2: [0.6, 0.4]
- Model 3: [0.8, 0.2]

**Average Probabilities:** [(0.7+0.6+0.8)/3, (0.3+0.4+0.2)/3] = [0.7, 0.3]

**Final Prediction:** 🐱 Cat (as it has the highest average probability).

---

#### 3. **Weighted Average Voting** ⚖️
Weighted Average Voting is similar to Averaging, but each model's prediction is assigned a weight based on its performance or importance. The final prediction is based on the weighted average of probabilities.

**Example:**
Three models predict probabilities for two classes (🐱 Cat and 🐶 Dog), with weights [0.5, 0.3, 0.2]:
- Model 1: [0.7, 0.3]
- Model 2: [0.6, 0.4]
- Model 3: [0.8, 0.2]

**Weighted Average Probabilities:**  
[(0.7×0.5 + 0.6×0.3 + 0.8×0.2), (0.3×0.5 + 0.4×0.3 + 0.2×0.2)] = [0.69, 0.31]

**Final Prediction:** 🐱 Cat (as it has the highest weighted average probability).
